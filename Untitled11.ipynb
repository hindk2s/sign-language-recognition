```python
import cv2
import numpy as np
import os
import time
import mediapipe as mp
from mediapipe.python.solutions import holistic
print('hi')
```

```python
mp_holistic = mp.solutions.holistic # Holistic model
mp_drawing = mp.solutions.drawing_utils # Drawing utilities
mp_drawing_styles = mp.solutions.drawing_styles

def mediapipe_detection(image, model):
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) # COLOR CONVERSION BGR 2 RGB
    image.flags.writeable = False                  # Image is no longer writeable
    results = model.process(image)                 # Make prediction
    image.flags.writeable = True                   # Image is now writeable 
    image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR) # COLOR COVERSION RGB 2 BGR
    return image, results

print('done')

def draw_landmarks(image, results):
    mp_drawing.draw_landmarks(
        image,
        results.face_landmarks,
        mp_holistic.FACEMESH_TESSELATION,  # Remplacer par FACEMESH_CONTOURS pour le flux vidéo
        landmark_drawing_spec=None,
        connection_drawing_spec=mp_drawing_styles.get_default_face_mesh_tesselation_style()  # Remplacer par get_default_face_mesh_contours_style() pour le flux vidéo
    ) # Draw face connections
    mp_drawing.draw_landmarks(
        image,
        results.pose_landmarks,
        mp_holistic.POSE_CONNECTIONS
    ) # Draw pose connections
    
    mp_drawing.draw_landmarks(
        image,
        results.left_hand_landmarks,
        mp_holistic.HAND_CONNECTIONS
    ) # Draw left hand connections
    mp_drawing.draw_landmarks(
        image,
        results.right_hand_landmarks,
        mp_holistic.HAND_CONNECTIONS
    ) # Draw right hand connections
print('done')

def draw_styled_landmarks(image, results):
    mp_drawing.draw_landmarks(
        image, results.face_landmarks, mp_holistic.FACEMESH_TESSELATION,
        mp_drawing.DrawingSpec(color=(80, 110, 10), thickness=1, circle_radius=1),
        mp_drawing.DrawingSpec(color=(80, 256, 121), thickness=1, circle_radius=1))

    mp_drawing.draw_landmarks(image, results.pose_landmarks, mp_holistic.POSE_CONNECTIONS,
                                    mp_drawing.DrawingSpec(color=(80, 22, 10), thickness=2, circle_radius=4),
                                    mp_drawing.DrawingSpec(color=(80, 44, 121), thickness=2, circle_radius=2)
                                    )

    mp_drawing.draw_landmarks(image, results.left_hand_landmarks, mp_holistic.HAND_CONNECTIONS,
                                    mp_drawing.DrawingSpec(color=(121, 22, 76), thickness=2, circle_radius=4),
                                    mp_drawing.DrawingSpec(color=(121, 44, 250), thickness=2, circle_radius=2)
                                    )

    mp_drawing.draw_landmarks(image, results.right_hand_landmarks, mp_holistic.HAND_CONNECTIONS,
                                    mp_drawing.DrawingSpec(color=(245, 117, 66), thickness=2, circle_radius=4),
                                    mp_drawing.DrawingSpec(color=(245, 66, 230), thickness=2, circle_radius=2)
                                    )
print('done') 

def extract_keypoints(results):
    pose = np.array([[res.x, res.y, res.z, res.visibility] for res in results.pose_landmarks.landmark]).flatten() if results.pose_landmarks else np.zeros(33*4)
    face = np.array([[res.x, res.y, res.z] for res in results.face_landmarks.landmark]).flatten() if results.face_landmarks else np.zeros(468*3)
    lh = np.array([[res.x, res.y, res.z] for res in results.left_hand_landmarks.landmark]).flatten() if results.left_hand_landmarks else np.zeros(21*3)
    rh = np.array([[res.x, res.y, res.z] for res in results.right_hand_landmarks.landmark]).flatten() if results.right_hand_landmarks else np.zeros(21*3)
    return np.concatenate([pose, face, lh, rh])

print('done')
```

```python
import os
import numpy as np
import cv2
from matplotlib import pyplot as plt

# Chemin vers les données exportées, tableaux numpy
DATA_PATH = os.path.join('MP2') 

# Récupération des noms des dossiers existants dans le répertoire MP
actions = os.listdir(DATA_PATH)

# Exemple de tableau res
res = [0.8, 0.6, 0.4, 0.9, 0.7, 0.5, 0.6, 0.3, 0.2, 0.1, 0.5]

# Troncature de la liste res pour correspondre à la longueur de actions
res = res[:len(actions)]

colors = [(245,117,16), (117,245,16), (16,117,245),(245,117,16), 
          (117,245,16), (16,117,245),(245,117,16), (117,245,16), 
          (16,117,245),(245,117,16), (117,245,16)]

def prob_viz(res, actions, input_frame, colors):
    output_frame = input_frame.copy()
    for num, prob in enumerate(res):
        cv2.rectangle(output_frame, (0,60+num*40), (int(prob*100), 90+num*40), colors[num % len(colors)], -1)
        cv2.putText(output_frame, actions[num], (0, 85+num*40), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 2, cv2.LINE_AA)
        
    return output_frame


image = np.zeros((720, 1280, 3), dtype=np.uint8)

plt.figure(figsize=(18,18))
plt.imshow(prob_viz(res, actions, image, colors))
plt.show()
```

```python
import tensorflow as tf
import cv2
import numpy as np
import mediapipe as mp
import torch
from transformers import GPTNeoForCausalLM, GPT2Tokenizer
import os
import time

# Charger le modèle GRU depuis le fichier enregistré
model_gru = tf.keras.models.load_model('TR.h5')

# Charger le tokenizer et le modèle GPT-Neo finement réglé
tokenizer = GPT2Tokenizer.from_pretrained("./gpt-neo-fine-tuned1")
model_gpt_neo = GPTNeoForCausalLM.from_pretrained("./gpt-neo-fine-tuned1").to("cuda" if torch.cuda.is_available() else "cpu")
max_new_tokens = 20

# Initialisez les variables nécessaires
sequence = []
sentence = []
predictions = []
threshold = 0.5
num_frames = 40  # Nombre de frames pour chaque séquence
cap = cv2.VideoCapture(0)  # Ouvrir la caméra
detected_words = set()  # Utilisez un set pour les mots détectés uniques

# Afficher le signal de démarrage
start_time = time.time()
while time.time() - start_time < 5:  # Afficher le message pendant 5 secondes
    ret, frame = cap.read()
    cv2.putText(frame, 'Detection starting in ' + str(int(5 - (time.time() - start_time))), (10, frame.shape[0] // 2), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2, cv2.LINE_AA)
    cv2.imshow('OpenCV Feed', frame)
    if cv2.waitKey(10) & 0xFF == ord('q'):
        cap.release()
        cv2.destroyAllWindows()
        exit()

# Boucle principale
with mp.solutions.holistic.Holistic(min_detection_confidence=0.5, min_tracking_confidence=0.5) as holistic:
    generated_sentence = ""  # Variable pour stocker la phrase générée
    while cap.isOpened():
        ret, frame = cap.read()

        image, results = mediapipe_detection(frame, holistic)
        
        draw_styled_landmarks(image, results)
        
        keypoints = extract_keypoints(results)
        sequence.append(keypoints)
        sequence = sequence[-num_frames:]
        
        if len(sequence) == num_frames:
            # Faire une prédiction avec le modèle GRU
            res_gru = model_gru.predict(np.expand_dims(sequence, axis=0))[0]
            predictions.append(np.argmax(res_gru))
            
            if np.unique(predictions[-10:])[0] == np.argmax(res_gru): 
                if res_gru[np.argmax(res_gru)] > threshold: 
                    detected_word = actions[np.argmax(res_gru)]
                    if detected_word not in detected_words:
                        detected_words.add(detected_word)
                        sentence.append(detected_word)

            if len(sentence) > 11: 
                sentence = sentence[-11:]

            image = prob_viz(res_gru, actions, image, colors)

            # Vérifiez si au moins cinq mots sont détectés
            if len(detected_words) >= 11:
                # Limitez la liste detected_words aux cinq derniers mots
                recent_words = list(detected_words)[-11:]

                # Concaténer les mots détectés en une seule phrase
                input_sentence = ' '.join(recent_words)

                # Créez un prompt pour guider la génération GPT-Neo
                prompt = "Form a short and meaningful sentence about our graduation day using all of the following words: " + input_sentence + ". Example: Thank you for your attention; we are happy to share this day with you; today we graduate; now we are engineers. Sentence:"

                # Tokenizez l'entrée pour GPT-Neo
                input_ids = tokenizer(prompt, return_tensors="pt").input_ids.to("cuda" if torch.cuda.is_available() else "cpu")

                # Générer la sortie avec GPT-Neo
                output = model_gpt_neo.generate(
                    input_ids,
                    max_length=input_ids.shape[1] + max_new_tokens,  # Permet jusqu'à max_new_tokens mots supplémentaires
                    num_beams=5,
                    num_return_sequences=1,
                    early_stopping=True,
                    no_repeat_ngram_size=2,
                    temperature=0.8
                )

                # Décoder et obtenir la phrase générée
                generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
                generated_sentence = generated_text[len(prompt):].strip()

        # # Effacer l'ancienne phrase générée en dessinant un rectangle noir
        # cv2.rectangle(image, (0, image.shape[0] - 60), (image.shape[1], image.shape[0]), (0, 0, 0), -1)  # Rectangle noir pour effacer l'ancienne phrasec

        # # Afficher la nouvelle phrase générée
        # cv2.putText(image, generated_sentence, (10, image.shape[0] - 20), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2, cv2.LINE_AA)
        
        # Ajuster la position et la taille du rectangle noir et du texte
        cv2.rectangle(image, (0, 0), (640, 60), (0, 0, 0), -1)  # Rectangle noir pour le fond du texte
        cv2.putText(image, ' '.join(sentence), (5, 50), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2, cv2.LINE_AA)  # Texte blanc
        
        cv2.imshow('OpenCV Feed', image)

        if cv2.waitKey(10) & 0xFF == ord('q'):
            break
            
    cap.release()
    cv2.destroyAllWindows()
```
